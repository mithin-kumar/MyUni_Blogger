```python
import pandas as pd

# Step 1: Extract Month
df['Month'] = df['Timestamp'].dt.month

# Step 2: Extract Day of the Week (0=Mon, 6=Sun)
df['DayOfWeek'] = df['Timestamp'].dt.dayofweek

# Step 3: Weekday or Weekend
df['IsWeekday'] = df['DayOfWeek'] < 5

# Step 4: Extract Hour of the Day
df['Hour'] = df['Timestamp'].dt.hour

# Step 5: Hour Category
def get_hour_category(hour):
    if 5 <= hour < 12:
        return 0
    elif 12 <= hour < 18:
        return 1
    else:
        return 2

df['HourCategory'] = df['Hour'].apply(get_hour_category)
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
import random

# Parameters
num_transactions = 10000
num_existing = int(0.8 * num_transactions)
num_new = num_transactions - num_existing

# Assume df is the original full transaction DataFrame
existing_customers = df['CustomerID'].unique()
selected_existing_customers = np.random.choice(existing_customers, size=num_existing, replace=True)

# Create new customer IDs
new_customers = [f"NEWCUST{str(i).zfill(4)}" for i in range(num_new)]

# Combine customer list
all_customers = np.concatenate([selected_existing_customers, new_customers])

# Shuffle the list
np.random.shuffle(all_customers)

# Generate OrderIDs
order_ids = [f"ORD{str(i).zfill(5)}" for i in range(num_transactions)]

# Generate random amounts
amounts = np.round(np.random.uniform(20, 500, size=num_transactions), 2)

# Generate timestamps in current month (July 2025)
start_date = datetime(2025, 7, 1)
timestamps = [start_date + timedelta(minutes=random.randint(0, 31*24*60)) for _ in range(num_transactions)]

# Simulate reward award and frequency (based on amount)
reward_awards = np.where(amounts > 150, np.random.uniform(10, 25, size=num_transactions), np.random.uniform(5, 10, size=num_transactions))
reward_freqs = np.where(amounts > 100, np.random.randint(5, 10, size=num_transactions), np.random.randint(0, 5, size=num_transactions))

# Create new DataFrame
new_df = pd.DataFrame({
    'CustomerID': all_customers,
    'OrderID': order_ids,
    'Amount': amounts,
    'Timestamp': timestamps,
    'RewardAward': np.round(reward_awards, 2),
    'RewardFrequency': reward_freqs
})
```

```python
df_ml['AmountBucket'] = pd.qcut(df['Amount'], q=4, labels=False)
df_ml['AvgMonthlySpend'] = df_ml.groupby('AmountBucket')['AvgMonthlySpend'].transform(lambda x: x.fillna(x.mean()))
df_ml['AvgMonthlyTxn'] = df_ml.groupby('AmountBucket')['AvgMonthlyTxn'].transform(lambda x: x.fillna(x.mean()))
df_ml
```

```python
df_ml['Timestamp'] = pd.to_datetime(df_ml['Timestamp'])
df_ml['Hour'] = df_ml['Timestamp'].dt.hour
df_ml['DayOfWeek'] = (df_ml['Timestamp'].dt.dayofweek )#> 5).astype(int)
df_ml['Day'] = df_ml['Timestamp'].dt.day

# Aggregate current month metrics
agg_df = df_ml.groupby('CustomerID').agg(
    TotalOrders=('OrderID', 'count'),
    TotalSpent=('Amount', 'sum'),
    AvgOrderValue=('Amount', 'mean'),
).reset_index()

df_ml = df_ml.merge(agg_df, on='CustomerID', how='left')
```

```python
features = ['AvgMonthlyTxn', 'AvgMonthlySpend', 'AmountBucket',
       'Hour', 'DayOfWeek', 'Day', 'TotalOrders', 'TotalSpent',
       'AvgOrderValue']

X = df_ml[features]
y1 = df_ml['RewardAward']
y2 = df_ml['RewardFrequency']
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
X_train, X_test, y1_train, y1_test = train_test_split(X, y1, test_size=0.2)
X_train2, X_test2, y2_train, y2_test = train_test_split(X, y2, test_size=0.2)

model_discount = RandomForestRegressor().fit(X_train, y1_train)
# model_discount = LinearRegression().fit(X_train, y1_train)
model_frequency = RandomForestRegressor().fit(X_train2, y2_train)
# model_frequency = LinearRegression().fit(X_train2, y2_train)
```

```python
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

seg_features = ['AvgMonthlyTxn', 'AvgMonthlySpend', 'AmountBucket',
       'HourCategory', 'DayOfWeek', 'Day', 'TotalOrders', 'TotalSpent',
       'AvgOrderValue']
scaler = StandardScaler()
X_seg = scaler.fit_transform(df_ml[seg_features])

kmeans = KMeans(n_clusters=3, random_state=42)
df_ml['CustomerSegment'] = kmeans.fit_predict(X_seg)
```

```python
df_ml
```

```python
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
# For reward discount
y1_pred = model_discount.predict(X_test)

# For reward frequency
y2_pred = model_frequency.predict(X_test2)
```

```python
mse_discount = mean_squared_error(y1_test, y1_pred)
rmse_discount = np.sqrt(mse_discount)
mae_discount = mean_absolute_error(y1_test, y1_pred)
r2_discount = r2_score(y1_test, y1_pred)

print("Reward Discount Prediction:")
print(f"  MAE : {mae_discount:.2f}")
print(f"  RMSE: {rmse_discount:.2f}")
print(f"  R²  : {r2_discount:.2f}")
```

```python
mse_frequency = mean_squared_error(y2_test, y2_pred)
rmse_frequency = np.sqrt(mse_frequency)
mae_frequency = mean_absolute_error(y2_test, y2_pred)
r2_frequency = r2_score(y2_test, y2_pred)

print("Reward Frequency Prediction:")
print(f"  MAE : {mae_frequency:.2f}")
print(f"  RMSE: {rmse_frequency:.2f}")
print(f"  R²  : {r2_frequency:.2f}")
```

```python
import tensorflow as tf
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Dense(64, activation='relu', input_shape=(len(features),)),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)  # One model for discount, another for frequency
])
model.compile(optimizer='adam', loss='mse')
model.fit(X_train, y1_train, epochs=20, validation_data=(X_test, y1_test))
```

```python
y_pred = model.predict(X_test).flatten()
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np

mse = mean_squared_error(y1_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y1_test, y_pred)
r2 = r2_score(y1_test, y_pred)

print("🔍 TensorFlow Model Evaluation (Regression):")
print(f"  MAE : {mae:.2f}")
print(f"  RMSE: {rmse:.2f}")
print(f"  R²  : {r2:.2f}")
```
